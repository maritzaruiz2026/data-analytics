## Permutations

```python
print(r_4)
```

```python
np.random.shuffle(r_4)
print(r_4)
```

The shuffle only randomized the row order when executed

```python
new_shuffle = np.random.permutation(r_4)
print(new_shuffle)
print(r_4)
```

Shuffle mdofifies the original array and the permutation leaves the original array alone. You can use shuffle when you don't need the original order and the permutation when you need to keep the original array intact.

## Grocery Problem.

```python
bag1 = np.array([2, 3, 4, 0])  
bag2 = np.array([4, 0, 1, 2])
print(bag1)
print(bag2)
```

```python
total_bag1 = np.sum(bag1)
total_bag2 = np.sum(bag2)
print(f'The first bag contains {total_bag1} items and the second bag contains {total_bag2} items.')
```

```python
total_categories = bag1 + bag2
veg, fruit, snacks, bev = total_categories
print(f'There are {veg} vegetables, {fruit} fruits, {snacks} boxes of snacks, and {bev} beverages in the two grocery bags')
```

## Teamwork

```python
#Ali reported knocking on 5 doors on Monday, 4 on Tuesday, 6 on Wednesday, 4 on Thursday, and 2 on Friday. Create a variable holding an array of these values.
ali = np.array([5, 4, 6, 4, 2])
```

```python
#Ben missed Monday because he was out of town. He reported knocking on 5 on Tuesday, 5 on Wednesday, 3 on Thursday, and 4 on Friday. Create a variable holding an array of Ben’s values.
ben = np.array([0, 5, 5, 3, 4])
```

```python
#Carla reported knocking on 3 doors on Monday, 6 on Tuesday, 3 on Thursday, and 4 on Friday. She was sick on Wednesday so missed that day. Create a variable holding an array of Carla’s values.
carla = np.array([3, 6, 0, 3, 4])
```

```python
#Compile the values reported by Ali, Ben, and Carla into a single array containing all door-knocks for the week, as all_knocks. Print the resulting array.
all_knocks = np.array([ali, ben, carla])
print(all_knocks)
```

```python
#Use all_knocks to calculate the total number of doors that each person knocked on for the week, and also the total number of doors that the group collectively knocked on. Print the results.
total_per_person = np.sum(all_knocks, axis=1)
total_group = np.sum(all_knocks)

print(total_per_person)
print(total_group)
```

```python
#The organization you’re working with estimates that for every two door-knocks, one of those people will actually turn out to vote. For your report to the organization, what is the rolling total of estimated votes that your group was able to turn out, day by day for last week? (Hint: Start with the rolling total by day, then multiply that array by an appropriate value to get the result.)
daily_totals = np.sum(all_knocks, axis=0)
rolling_knocks = np.cumsum(daily_totals)
estimated_votes = rolling_knocks * 0.5
results = np.vstack([daily_totals, rolling_knocks, estimated_votes])
print(results)

```
