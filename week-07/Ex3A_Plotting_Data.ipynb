# Exercise 3.A Getting familiar with Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Creating Basic Plots.

```python
x = [1, 2, 3, 4, 5] 
y = [2, 3, 5, 7, 11]
```

```python
plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,y, marker='o')
plt.show()
```

```python
plt.plot(x, y, marker='o')
plt.title("Example Graph")
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.show()
```

## Styling Plots

```python
plt.plot(x, y, linestyle='--', color='red')
plt.show()
```

```python
plt.plot(x, y, linestyle='--',marker='s', markersize=10, color='green')
plt.show()
```

```python
plt.plot(x, y, linestyle='--',marker='D', markersize=10, color='Orange')
plt.show()
```

```python
plt.plot(x, y, linestyle='--',marker='*', markersize=10, color='purple')
plt.show()
```

# Exercise 3.B More plotting practice

## Advanced Plotting Techniques.

```python
products = ['Product A', 'Product B', 'Product C', 'Product D'] 
sales = [23, 17, 35, 29]
```

```python
plt.bar(products, sales, color='skyblue', width=0.5)
plt.title("Sales by Product")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.show()
```

```python
x = np.random.rand(50) 
y = np.random.rand(50)
plt.scatter(x,y)
```

```python
markersize = np.random.rand(50) * 100
plt.scatter(x, y, s=markersize, cmap='inferno')
plt.show()
```

## Subplots

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)

histogram = np.random.randn(1000)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5))
ax1.plot(x, y)
ax1.set_title("Wave")
ax1.set_xlabel("X Values")
ax1.set_ylabel("sin(x)")

ax2.hist(histogram, bins=30, color='orange', edgecolor='red')
ax2.set_title("Histogram of Random Numbers")
ax2.set_xlabel("Value")
ax2.set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```
