## Generating Random Numbers

```python
r_l = np.random.randint(1,20, size=2)
print(r_l)
```

```python
r_2 = np.random.randint(1, 21, size=(1, 2))
print(r_2)
```

```python
r_3 = np.random.rand(2, 1)
print(r_3)
```

```python
r_4 = np.random.randint(20, 101, size=(4, 5))
print(r_4)
```

```python
r_5 = np.random.rand(4, 5)
print(r_5)
```

## Generating Random Choices from an Array

```python
students = np.array(['Onur', 'Longley', 'Jessica', 'Eyerusalem', 'Godknows', 'Rodas', 'Will', 'Kiki', 'Sultan', 'Cagla', 'Paola', 'Mina'])
```

```python
call_on = np.random.choice(students)
print(call_on)
```

```python
breakout_groups = np.random.choice(students, size=(3, 4))
print(breakout_groups)
#This wouldn't be a good way to generate breakout groups because some names are repetititve
```

```python
call_on_weighted = np.random.choice(students, size=5, p=[0.30,0.10,0.05,0.05,0.00,0.10,0.10,0.10,0.05,0.05,0.00,0.10])
print(call_on_weighted)
#I randomized the probability decimals, I kept gettting errors at first because they did not equal 1, so had to keep track of my math
```

## Generating a Normal Distribution

```python
bell_curve = np.random.normal(loc=10, scale=5, size=(4, 5))
print(bell_curve)
```
