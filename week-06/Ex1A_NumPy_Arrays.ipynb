# Looping Through Arrays

```python
arrA = np.array([[11, 12, 13],[14, 15, 16]])
print(arrA)
```

```python
for x in arrA:
    print(x * 2)
```

```python
for x in arrA:
    for y in x:
        print(y * 2)
```

In the first loop all of the rows were multiplied by 2 and executed in the original row format. The second loop prints every individual element by 2 and returns them individually. 

```python
for x in arrA:
    for y in x:
        loop_result = x * 2
    print(loop_result) #This multiplies the entire row x by 2. The inner loop "for y in x:" runs through each element (y) but is never used. The nested loop does not affect the excuted result. This reult is the same to:
    #for x in arrA:
        #print(x*2)
```

## Joining and Stacking Arrays

```python
joined_array_1 = np.concatenate ((arrA, arrA))
print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA, arrA), axis=1)
print(joined_array_2)
```

```python
stacked_array_1 = np.stack((joined_array_1, joined_array_1))
print(stacked_array_1)
```

```python
stacked_array_2 = np.stack((joined_array_1, joined_array_2), axis=2)
#Got error code saying all aways must have the same shape in order to stack them
```

```python
another_joined_array = np.concatenate((joined_array_1, arrA), axis=0)
print(another_joined_array)
```

```python
another_2 = np.concatenate((joined_array_1, arrA), axis=1)
#Got an error that said "all the input array dimensions except for the concatenation axis must match exactly" so they were not able to be joined due to not having the same # of rows to concatenate on axis = 1
```

```python
var1 = np.hstack((arrA, arrA))
print(var1)   #This executes a horizontal stack, side by side, so it adds columns along its axis
#[[11 12 13 11 12 13]
# [14 15 16 14 15 16]]

var2 = np.vstack((arrA, arrA))
print(var2)   #This executes a vertical stack, on top of each other, so it adds rows along the row axis
#[[11 12 13]
# [14 15 16]
# [11 12 13]
# [14 15 16]]

var3 = np.dstack((arrA, arrA))
print(var3)   #This executes a depth stack, so it puts arrays on top of each other and groups like elements
#[[[11 11]    #So here 11 is grouped
#  [12 12]
#  [13 13]]

# [[14 14]
#  [15 15]
#  [16 16]]]
```
